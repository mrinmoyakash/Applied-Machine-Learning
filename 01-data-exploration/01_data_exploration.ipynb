## 1.4 Slicing and Extracting Data

Isolating a single column returns a pandas Series.

```
# Isolating a single column
df['Name'].head()
```

Passing a list of column names returns a DataFrame with just those columns.

```
# Isolating multiple columns
df[['Name', 'Sex', 'Age']].head()
```

`.loc[]` selects rows by their index label. Here it selects the single row labeled `5`.

```
# Isolating a single row
df.loc[5]
```

`.loc[]` also accepts a range of index labels. Note that unlike Python slicing, this range is inclusive of both endpoints.

```
# Isolating multiple rows with range
df.loc[5:20]
```

`.loc[]` accepts a list of index labels too, in any order.

```
# Isolating multiple rows with list
df.loc[[1, 5, 10, 6]]
```

Make a copy of a slice so that later edits don't affect the original `df`.

```
dft = df.loc[5:20].copy()
dft.head()
```

`dft` only contains rows labeled 5 through 20, so label `0` doesn't exist in it anymore. `.loc[0]` looks up by *label*, so this raises a `KeyError` — caught here by the `try/except`.

```
try:
    print(dft.loc[0])
except:
    print("There is an error")
```

`.iloc[0]` looks up by *positional* index instead, so it works fine here — it simply returns the first row of `dft`, regardless of its label.

```
try:
    print(dft.iloc[0])
except:
    print("There is an error")
```

`.iloc[]` can select a row range and a column range together, using positions.

```
# Isolating both rows and columns with range
# loc[] doesn't work with column range but iloc[] does
df.iloc[0:20, 2:5].head()
```

`.loc[]` can select a row range together with a list of column *names* (rather than positions).

```
# Isolating both rows and columns with column names in list
# loc[] works but .iloc[] doesn't
df.loc[0:20, ['Name', 'Sex', 'Age']].head()
```

You can assign directly into `.loc[row, column]` to edit a specific cell of the DataFrame.

```
# That's how we change specific attribute of a single row
dft.loc[5, 'Name'] = 'Ratri'

# Change these attributes
dft.loc[5, 'Sex'] = 'female'
dft.loc[5, 'Age'] = 26
dft.loc[5, 'Survived'] = 1
dft.head()
```

`.sort_values(by=...)` sorts the DataFrame by a given column.

```
# We can sort the data by an attribute
dft.sort_values(by='Age', ascending=True).head()
```

## 1.5 Slicing with Conditions

Pandas slicing with conditions lets you select a subset of rows based on a criterion. `.loc[]` combined with a boolean condition (a *boolean mask*) is one common way to filter data. (There's also the `.query()` method, which we won't cover here.)

`df['Age'] < 18` produces a boolean mask that is `True` for every underage passenger; `.loc[]` then keeps only those rows.

```
# Let's find the underage people in the data set
df.loc[df['Age']<18].head()
```

Combine multiple conditions with `&` (and) or `|` (or). Each condition needs its own parentheses.

```
# Now let's find underage people who are also female
df.loc[ (df['Age']<20) & (df['Sex']=='female')].head()
```

## 1.6 Cleaning Data

### 1.6.1 Dropping Missing Values

Copy the DataFrame first so cleaning steps don't overwrite the original `df`.

```
# We will copy the dataframe first in order to not change the original
dft = df.copy()
dft.head()
```

Check which columns have missing values before cleaning.

```
dft.isnull().sum()
```

`.dropna()` removes every row that has at least one missing value.

```
# Droppping missing values
dft = dft.dropna()
dft.head()
```

`.dropna(axis=1)` instead removes every *column* that has at least one missing value.

```
dft = df.copy()
dft = dft.dropna(axis=1)
dft.head()
```

### 1.6.2 Replacing Missing Values

Take a fresh look at the full dataset before filling in missing values.

```
df
```

Find the rows where `Age` is missing.

```
dft = df.copy()
dft.loc[dft['Age'].isnull()].head()
```

Compute the average age, which we'll use to fill in the missing values.

```
dft['Age'].mean()
```

`.fillna(value)` replaces every missing value in the DataFrame with `value` — here, the mean age. Afterward, no rows should have a missing `Age`.

```
# Get the mean of age
mean_value = dft['Age'].mean()
# Fill missing values using .fillna()
dft = dft.fillna(mean_value)
dft.loc[dft['Age'].isnull()].head()
```

Confirm there are no more missing `Age` values (this should return an empty DataFrame).

```
dft.loc[dft['Age'].isnull()].head()
```

### 1.6.3 Removing Duplicates

Start again from a fresh copy of the original data.

```
dft = df.copy()
dft.head()
```

`.drop_duplicates()` removes rows that are exact duplicates of an earlier row.

```
dft = dft.drop_duplicates()
dft.head()
```

## 1.7 Manipulating Columns

`.rename(columns={...})` renames one or more columns.

```
# We can rename a column using the rename() method
dft = dft.rename(columns = {'Parch':'Porch'})
dft.head()
```

You can create a new column by assigning to it directly. Here we start every passenger as `baby = 1`, then use a conditional `.loc[]` assignment to set `baby = 0` for anyone older than 10.

```
# Now we will add a new column named baby
# baby is a boolean value column (1 or 0)
# It will show whether a passenger was a baby or not
dft['baby'] = 1
dft.loc[dft['Age']>10, 'baby'] = 0
dft.head()
```

## 1.8 Further Analysis

`.mean()` on a column computes its average. Pandas also offers `.median()` and `.mode()` for the other common measures of central tendency.

```
# We can find mean, median, mode
df['Age'].mean()
```

`.groupby(column)` groups rows that share the same value in `column`, so you can then aggregate within each group — here, the average age separately for each sex.

**Note:** the original version of this cell used lowercase `'sex'`/`'age'`, but the actual column names in this dataset are capitalized (`'Sex'`, `'Age'`) — pandas column lookups are case-sensitive, so that raised a `KeyError`. Fixed below.

```
# Find mean of age of females and males individually
df.groupby('Sex')['Age'].mean()
```
